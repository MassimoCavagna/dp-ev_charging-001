```python
import json
from utils.config import load_ocm_raw_json_path
from pyspark.sql import functions as F
```

```python
RUN_ID = 1793953962649
OCM_RAW_JSON_PATH = load_ocm_raw_json_path()
FILE_NAME = f"pois_italy_{RUN_ID}.json"
PROJECT_ID = '`dp-ev_charging-001`'
TARGET_TABLE = 'bronze.ocm_poi'
```

```python


ocm_poi_df = (
        spark.read.json(OCM_RAW_JSON_PATH + FILE_NAME, multiLine=True)
        .withColumns({
                    "_run_id": F.lit(RUN_ID),
                    "_source_file_name": F.lit(FILE_NAME),
                    "_ingested_at": F.current_timestamp()
                    })
)
```

```python
spark.sql(f"""DELETE FROM {PROJECT_ID}.{TARGET_TABLE} WHERE _run_id = {RUN_ID}""")
```

```python

ocm_poi_df.write.mode("append").saveAsTable(f"{PROJECT_ID}.{TARGET_TABLE}")
```
